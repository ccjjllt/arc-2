# Generalized local CPU repair on frozen D24 programs

```python
import base64, hashlib, json, subprocess, sys, zlib
from pathlib import Path
work = Path('/kaggle/working/020-cpu-code')
work.mkdir(exist_ok=True)
blobs = json.loads(zlib.decompress(base64.b64decode('')))
expected = {'replay_local_repair.py': '23ed680e02f1057c3e63bd628e569e02cfc677b06b5526f2042ab8bf57292835', 'solve_program.py': '405e2cda3226ae1aca6a6ff762dbd5a2bdec1cba3510b97d335072cd41c73e88', 'local_repair.py': '0de8ad24ba02de9c1deefcd762a2c763676d5413fb484b10e085411dc7c584c0', 'submit_lib.py': '53ded4ec6623a68c25cbf9041de42fad8f5b3f8ffcde253dfd1afe147ece1b0e', 'input.json': '6ec1ea168320a7d06ed29c830313fcd98ee8612dd166ec0faaaede0fe36f7d21'}
for (name, value) in blobs.items():
    raw = base64.b64decode(value)
    assert hashlib.sha256(raw).hexdigest() == expected[name]
    (work / name).write_bytes(raw)
print(json.dumps({'event': 'CPU_REPAIR_BEGIN', 'source_sha256': expected}), flush=True)
subprocess.run([sys.executable, '-u', str(work / 'replay_local_repair.py'), '--input', str(work / 'input.json'), '--output', '/kaggle/working/cpu-replay', '--workers', '4'], cwd=str(work), timeout=1500, check=True)
```

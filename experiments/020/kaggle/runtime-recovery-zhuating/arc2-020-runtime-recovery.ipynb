# Exact public runtime recovery
Private CPU carrier for zhuating. Internet is used only to download the original public wheel files and verify their recorded hashes. No competition, model, or dataset input is attached; downloaded code is not installed or executed.


```python
import base64, hashlib, json, subprocess, sys, zlib
from pathlib import Path
repo = Path('/kaggle/working/020-restore-repo')
output = Path('/kaggle/working/qwen38-vllm0272-cu130-wheelhouse-v1')
blobs = json.loads(zlib.decompress(base64.b64decode('')))
expected = {'experiments/020/restore_wheelhouse.py': '2589c831bb69461ac93555e2778653915b32d59d7483f4e09fd3627920f835f7', 'experiments/014/public-sources/qwen38-vllm0272-manifest.json': 'c9a31a531c2ec9e703de936859906a2db2a78a8987419de744e2bd3c0e2861e1', 'experiments/014/public-sources/qwen38-vllm0272-requirements.txt': 'ebb525ba1fda0bc955fc19fe06fa381dd994e5b9a26a6e154a7ecd923f85d133'}
for relative_path, value in blobs.items():
    raw = base64.b64decode(value)
    assert hashlib.sha256(raw).hexdigest() == expected[relative_path], relative_path
    target = repo / relative_path
    target.parent.mkdir(parents=True, exist_ok=True)
    target.write_bytes(raw)
print(json.dumps({'event': 'CPU_EXACT_RUNTIME_RECOVERY_BEGIN', 'source_sha256': expected, 'gpu': False, 'competition_sources': []}), flush=True)
result = subprocess.run([sys.executable, '-u', str(repo / 'experiments/020/restore_wheelhouse.py'), '--output', str(output), '--workers', '8'], cwd=str(repo), timeout=7200)
print(json.dumps({'event': 'CPU_EXACT_RUNTIME_RECOVERY_EXIT', 'returncode': result.returncode}), flush=True)
assert result.returncode == 0, 'Exact runtime recovery failed; do not consume this version'
report = json.loads((output / 'restore-report.json').read_text())
assert report['status'] == 'complete_exact_original_wheels'
assert report['verified_wheels'] == report['expected_wheels'] == 193
assert report['failed_wheels'] == 0
assert report['verified_wheel_bytes'] == 4900988522
assert len(list((output / 'wheels').glob('*.whl'))) == 193
assert hashlib.sha256((output / 'requirements-qwen38-runtime.txt').read_bytes()).hexdigest() == 'ebb525ba1fda0bc955fc19fe06fa381dd994e5b9a26a6e154a7ecd923f85d133'
print(json.dumps({'event': 'CPU_EXACT_RUNTIME_RECOVERY_READY', 'status': report['status'], 'verified_wheels': report['verified_wheels'], 'verified_wheel_bytes': report['verified_wheel_bytes'], 'output': str(output)}), flush=True)
```
